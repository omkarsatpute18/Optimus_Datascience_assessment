 Data Cleaning Decisions

Based on the analysis, the following principles will be used:

- The identifier will not be treated as a predictive feature.
- Exact duplicate records will be investigated before removal.
- Duplicate identifiers will be investigated separately from exact duplicates.
- Numerical missing values will not automatically be replaced without considering
  the meaning of the variable.
- Categorical missing values will be treated according to the meaning of
  missingness and the modeling requirements.
- Variables with extremely high missingness will be evaluated individually.
- Variables that violate the May 2014 information cutoff will not be treated
  as available information.

Key EDA Findings

### Dataset
- The dataset contains approximately 50,000 voter records.
- The dataset contains 39 variables.
- Variables include demographics, voter history, and historical precinct turnout.

### Data Quality
- Several variables have substantial missingness.
- Historical voter-history variables have no missing observations.
- Precinct turnout variables have extremely low missingness.
- Duplicate records and identifiers were explicitly checked.

### Historical Participation
- Historical participation rates vary across elections.
- Historical general and primary participation show different patterns.
- Historical participation variables are correlated with one another.

### Temporal Considerations
- The May 2014 cutoff must be respected.
- Variables representing information unavailable by that date should not be used
  as available information.

```python
# Before Preparaing the Model our needs to clean
# DATA PREPARATION
# Comment: 1,141 exact duplicate rows (all duplicate optimus_ids are exact copies).
# Remove them so each voter appears once and train/validation splits cannot share copies.
orig_cols = [c for c in df.columns if c not in ("historical_participation_count", "age_group")]
assert df[orig_cols].duplicated().sum() == df["optimus_id"].duplicated().sum()  # copies only, no conflicts

model_df = df.drop_duplicates(subset="optimus_id").reset_index(drop=True)
print("Rows before:", len(df), "| after:", len(model_df))
assert model_df["optimus_id"].is_unique
```

```python
# Comment: congressional district is a label, not a quantity, so treat it as categorical.
model_df["cd"] = pd.Categorical(model_df["cd"])

# Comment: static features kept (<=62% missing). Variables >90% missing are dropped.
STATIC = ["party", "ethnicity", "income", "education", "maritalstatus",
          "dwellingtype", "net_worth", "home_owner_or_renter", "cd", "dma"]
for c in STATIC:
    if model_df[c].dtype == "object":
        model_df[c] = model_df[c].astype("category")
# Note: income already contains an "Unknown" level, which stays as its own category.
```

```python
# MODEL SELECTION AND TUNING
from scipy.stats import loguniform
from sklearn.model_selection import StratifiedKFold, cross_validate, RandomizedSearchCV, cross_val_predict
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import brier_score_loss, log_loss

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
```

```python
#FEATURE CONSTRUCTION, TRAINING FRAME AND TRAIN/VALIDATION SPLIT
from sklearn.model_selection import train_test_split

def build_features(data, target_year):
    """Comment: builds lag-based features using ONLY information from before target_year.
    The same-year primary (vh14p) is excluded because the June 2014 primary falls after
    the May 2014 cutoff."""
    X = pd.DataFrame(index=data.index)
    for lag in LAGS:
        yy = f"{(target_year - lag) % 100:02d}"
        X[f"gen_lag{lag}"] = data[f"vh{yy}g"]
        X[f"pri_lag{lag}"] = data[f"vh{yy}p"]
    base = f"{(target_year - 2) % 100:02d}"
    X["prec_g_lag2"] = data[f"g{base}_precinct_turnout"]
    X["prec_p_lag2"] = data[f"p{base}_precinct_turnout"]
    X["age_at_target"] = data["age"] - (SCORE_YEAR - target_year)
    X["n_prior_general"] = X[[f"gen_lag{l}" for l in LAGS]].sum(axis=1)
    X["n_prior_primary"] = X[[f"pri_lag{l}" for l in LAGS]].sum(axis=1)
    for c in STATIC:
        X[c] = data[c]
    return X

# Training frame: target = 2010 general election
X_train_all = build_features(model_df, TRAIN_YEAR)
y_all = model_df["vh10g"]

# Comment: keep only people old enough to vote in 2010 (rows with missing age are excluded here)
eligible = X_train_all["age_at_target"] >= 18
X_tr, y_tr = X_train_all[eligible], y_all[eligible]

# Split: 80% to fit, 20% held back as the final exam
X_fit, X_val, y_fit, y_val = train_test_split(
    X_tr, y_tr, test_size=0.2, stratify=y_tr, random_state=RANDOM_STATE)

print("Training rows:", len(X_tr), "| fit:", len(X_fit), "| validation:", len(X_val))
print("2010 general turnout rate:", round(y_tr.mean(), 4))
print("Features:", list(X_tr.columns))

# Leakage guard
banned = {"historical_participation_count", "age_group", "vh14p", "vh10g", "optimus_id"}
assert banned.isdisjoint(X_tr.columns), banned & set(X_tr.columns)
print("No post-cutoff or target columns in the feature set.")
```

```python
# Candidate 1: logistic regression (simple, interpretable benchmark)
cat_cols = X_fit.select_dtypes("category").columns.tolist()
num_cols = [c for c in X_fit.columns if c not in cat_cols]

pre = ColumnTransformer([
    ("num", Pipeline([("imp", SimpleImputer(strategy="median", add_indicator=True)),
                      ("sc", StandardScaler())]), num_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore", min_frequency=50), cat_cols)])
logit = Pipeline([("pre", pre), ("clf", LogisticRegression(max_iter=1000))])

# Candidate 2: gradient boosting (your current model)
hgb = HistGradientBoostingClassifier(
    categorical_features="from_dtype", max_depth=4, learning_rate=0.05,
    max_iter=300, early_stopping=True, random_state=RANDOM_STATE)

scoring = {"auc": "roc_auc", "brier": "neg_brier_score", "logloss": "neg_log_loss"}
rows = []
for name, mdl in [("Logistic regression", logit), ("Gradient boosting", hgb)]:
    r = cross_validate(mdl, X_fit, y_fit, cv=cv, scoring=scoring, n_jobs=-1)
    rows.append({"model": name,
                 "AUC": r["test_auc"].mean(), "AUC_std": r["test_auc"].std(),
                 "Brier": -r["test_brier"].mean(), "LogLoss": -r["test_logloss"].mean()})
cv_table = pd.DataFrame(rows).round(4)
print(cv_table)
```

```python
param_dist = {
    "max_depth": [3, 4, 5, 6, None],
    "max_leaf_nodes": [15, 31, 63],
    "learning_rate": loguniform(0.01, 0.2),
    "max_iter": [200, 300, 500],
    "min_samples_leaf": [20, 50, 100, 200],
    "l2_regularization": loguniform(1e-3, 10),
}
search = RandomizedSearchCV(
    hgb, param_dist, n_iter=25, scoring="roc_auc", cv=cv,
    random_state=RANDOM_STATE, n_jobs=-1, verbose=1)
search.fit(X_fit, y_fit)

print("Best CV AUC:", round(search.best_score_, 4))
print("Best parameters:", search.best_params_)
best_hgb = search.best_estimator_
```

```python
oof = cross_val_predict(best_hgb, X_fit, y_fit, cv=cv, method="predict_proba", n_jobs=-1)[:, 1]

grid = np.arange(0.20, 0.81, 0.01)
f1s = [f1_score(y_fit, (oof >= t).astype(int)) for t in grid]
THRESHOLD = float(grid[int(np.argmax(f1s))])
print("Chosen threshold (max F1 on out-of-fold):", round(THRESHOLD, 2))

plt.plot(grid, f1s); plt.axvline(THRESHOLD, ls="--")
plt.xlabel("Threshold"); plt.ylabel("F1"); plt.title("Threshold selection (training folds)"); plt.show()
```

```python
# Comment: calibration makes vote_prob behave like a real probability.
calibrated = CalibratedClassifierCV(best_hgb, method="isotonic", cv=5)
calibrated.fit(X_fit, y_fit)

def report(name, p):
    pred = (p >= THRESHOLD).astype(int)
    return {"model": name,
            "AUC": roc_auc_score(y_val, p), "Brier": brier_score_loss(y_val, p),
            "Accuracy": accuracy_score(y_val, pred), "Precision": precision_score(y_val, pred),
            "Recall": recall_score(y_val, pred), "F1": f1_score(y_val, pred)}

base_model = hgb.fit(X_fit, y_fit)                     # your original settings
results = pd.DataFrame([
    report("Original boosting", base_model.predict_proba(X_val)[:, 1]),
    report("Tuned boosting", best_hgb.predict_proba(X_val)[:, 1]),
    report("Tuned + calibrated", calibrated.predict_proba(X_val)[:, 1]),
]).round(4)
print(results)
print(confusion_matrix(y_val, (calibrated.predict_proba(X_val)[:, 1] >= THRESHOLD).astype(int)))
```

# Conclusion

The dataset was systematically examined for structure, data types,
missing values, duplicate records, historical voting information,
precinct-level turnout information, and temporal availability.

The exploratory analysis provides an understanding of the historical
patterns present in the dataset and identifies variables requiring
special treatment.

A key consideration is the May 2014 prediction cutoff. Therefore,
the predictive methodology must distinguish information that would
have been available at that point from information that became
available later.

Before any predictive modeling is performed, a historical validation
strategy should be established and documented so that the methodology
can be evaluated using appropriate historical information.
